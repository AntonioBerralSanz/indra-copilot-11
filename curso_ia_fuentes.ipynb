# Fuentes del curso: IA, prompting y sesgos

Este cuaderno recoge materiales recomendados para profundizar en dos bloques clave del curso: frameworks de prompting y sesgos en la IA. Se puede ampliar con referencias específicas del profesor o de la bibliografía oficial del programa.


## 1) Material sobre frameworks de prompting

Los frameworks de prompting ayudan a estructurar instrucciones para que un modelo responda de forma más robusta, fiel y útil. A continuación se incluyen referencias clave:

- OpenAI. Prompt Engineering Guide. https://platform.openai.com/docs/guides/prompt-engineering
- Anthropic. Prompt Engineering Overview. https://docs.anthropic.com/en/docs/build-with-claude/prompt-engineering/overview
- DeepLearning.AI. ChatGPT Prompt Engineering for Developers. https://www.deeplearning.ai/courses/chatgpt-prompt-engineering-for-developers/
- LangChain. Prompt Templates and Chains. https://python.langchain.com/docs/modules/prompts/
- Google AI. Prompt design strategies. https://ai.google.dev/gemini-api/docs/prompting-strategies

### Ideas clave
- Definir contexto, objetivos y formato esperado.
- Separar instrucciones, ejemplos y restricciones.
- Usar técnicas como Few-Shot, Chain-of-Thought, role prompting y structured output.
- Evaluar la calidad con pruebas y criterios concretos.


## 2) Material sobre sesgos en la IA

La IA puede reproducir sesgos presentes en datos, diseño o decisiones humanas. Estas fuentes permiten entender el problema y proponer mitigaciones:

- NIST. AI Risk Management Framework. https://www.nist.gov/itl/ai-risk-management-framework
- Google PAIR. Responsible AI Practices. https://pair.withgoogle.com/
- MIT Media Lab / Joy Buolamwini. Gender Shades: Intersectional Accuracy Disparities in Commercial Gender Classification. https://proceedings.mlr.press/v81/buolamwini18a.html
- UNESCO. Recommendation on the Ethics of Artificial Intelligence. https://www.unesco.org/en/artificial-intelligence/recommendation-ethics
- IBM. AI Fairness 360. https://aif360.readthedocs.io/

### Ideas clave
- El sesgo puede aparecer en los datos, en la representación y en la validación.
- La equidad no es equivalente a la igualdad; hay que medir impacto real.
- Es necesario evaluar rendimiento por grupos, documentar supuestos y monitorizar resultados.
- La transparencia, la trazabilidad y la auditoría son esenciales.

# Productos de estudio: resumen, preguntas y glosario

## 1) Resumen del curso (máximo 10 puntos)

1. La IA generativa responde a instrucciones en lenguaje natural, pero necesita contexto, estructura y criterios claros para ofrecer resultados útiles.
2. El prompting es una disciplina práctica: la calidad de la respuesta mejora cuando se especifican objetivo, formato, tono y restricciones.
3. Los frameworks de prompting ayudan a organizar la interacción en etapas: instrucción, contexto, ejemplos, salida esperada y validación.
4. El uso de ejemplos (few-shot) y el establecimiento de roles pueden mejorar la calidad y la consistencia de la respuesta.
5. La IA no tiene ‘comprensión’ humana plena; genera respuestas a partir de patrones estadísticos y se debe evaluar su fiabilidad.
6. Las alucinaciones aparecen cuando el modelo produce información que parece válida pero no está respaldada por fuentes fiables.
7. Los sesgos en la IA pueden originarse en datos, decisiones de diseño, etiquetas humanas o desigualdades del entorno de entrenamiento.
8. La equidad y la transparencia requieren auditoría, evaluación por grupos y monitorización continua, no solo mejoras técnicas aisladas.
9. La ética de la IA combina rendimiento técnico con responsabilidad social: impacto, trazabilidad, límites y uso adecuado.
10. La competencia clave del curso es aprender a usar la IA con criterio, no solo con entusiasmo: pedir mejor, verificar mejor y comunicar mejor.

## 2) Preguntas de repaso con respuesta

1. ¿Qué es el prompting?
   - Respuesta: Es el conjunto de instrucciones y contexto que se le da a un sistema de IA para orientar su respuesta.

2. ¿Por qué es importante definir el formato esperado?
   - Respuesta: Porque reduce ambigüedad y ayuda a obtener resultados más estructurados, comparables y útiles.

3. ¿Qué diferencia hay entre zero-shot y few-shot prompting?
   - Respuesta: Zero-shot significa pedir una tarea sin ejemplos; few-shot significa proporcionar unos ejemplos antes de la tarea final.

4. ¿Qué es una alucinación en IA?
   - Respuesta: Es una respuesta que parece plausible pero contiene información falsa, inventada o sin soporte fiable.

5. ¿Qué es un sesgo algorítmico?
   - Respuesta: Es un error o tendencia sistemática que hace que el sistema favorezca o perjudique a ciertos grupos o situaciones de forma injusta.

6. ¿De dónde puede venir un sesgo en IA?
   - Respuesta: De datos históricos, etiquetas humanas, muestras no representativas, decisiones de diseño o contextos de uso desiguales.

7. ¿Qué significa evaluar una respuesta de IA?
   - Respuesta: Significa comprobar si es correcta, útil, segura, completa y alineada con la intención original del usuario.

8. ¿Cuál es la relación entre IA y responsabilidad ética?
   - Respuesta: La IA debe usarse con supervisión, transparencia y criterio humano; la calidad técnica no elimina la necesidad de evaluar su impacto social.

## 3) Glosario de 10 conceptos

- Prompt: Instrucción o conjunto de indicaciones que se ofrece al modelo para orientar la respuesta.
- Contexto: Información adicional que ayuda al modelo a situarse y responder con más precisión.
- Zero-shot prompting: Solicitar una tarea sin ejemplos previos.
- Few-shot prompting: Incluir ejemplos para guiar el patrón deseado de respuesta.
- Hallucination: Generación de contenido falso o no verificado que se presenta como verdadero.
- Bias (sesgo): Tendencia sistemática que afecta la justicia o la representatividad de una salida.
- Robustez: Capacidad del sistema para mantener un buen rendimiento ante cambios o entradas diferentes.
- Structured output: Respuesta con formato definido, como tabla, JSON, lista o esquema concreto.
- Evaluación: Proceso de valorar la calidad, precisión y relevancia de una respuesta generada.
- Ética de la IA: Conjunto de principios para asegurar un uso responsable, justo y transparente.

## 4) Ampliación: mapa conceptual

El curso puede resumirse con esta relación:

Prompting -> contexto + instrucciones + formato + ejemplos -> respuesta más útil
IA generativa -> fortalezas + riesgos -> necesidad de evaluación
Riesgos -> alucinaciones + sesgos + falta de supervisión -> responsabilidad ética

En otras palabras: una IA bien guiada y evaluada puede ser muy útil, pero requiere criterio humano, comprobación y conciencia crítica.
